```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df = pd.read_csv("netflix_movies_detailed_up_to_2025.csv")


df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()


df = df.dropna(subset=['release_year', 'type'])
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

```python
plt.figure(figsize=(10,5))
df['release_year'].value_counts().sort_index().plot()
plt.title("Content Released Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(8,5))
top_countries.plot(kind='bar')
plt.title("Top 10 Countries Producing Content")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(y='rating', data=df, order=df['rating'].value_counts().index)
plt.title("Content Rating Distribution")
plt.show()
```

```python
# Clean column names
df.columns = df.columns.str.strip()

# Check column exists
if 'duration' not in df.columns:
    print("❌ 'duration' column not found")
else:
    print("✅ duration column found")

# Show sample values
print(df['duration'].dropna().head(10))

# Convert to string
df['duration'] = df['duration'].astype(str)

# Extract numbers
df['duration_num'] = df['duration'].str.extract(r'(\d+)')

# Convert to numeric
df['duration_num'] = pd.to_numeric(df['duration_num'], errors='coerce')

# Check valid values
valid_count = df['duration_num'].notnull().sum()
print("Valid duration values:", valid_count)

# If no data → stop
if valid_count == 0:
    print("❌ No valid duration data found. Dataset issue.")
else:
    plt.figure(figsize=(8,5))
    sns.histplot(df['duration_num'].dropna(), bins=30)
    plt.title("Duration Distribution")
    plt.show()
```

```python
# Top Release Years
plt.figure(figsize=(10,5))
df['release_year'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Release Years")
plt.show()
```

```python
# Only numeric columns
numeric_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(6,4))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
# Clean column names (VERY IMPORTANT)
df.columns = df.columns.str.strip()

# Check if column exists
if 'listed_in' in df.columns:

    # Drop null values
    genre_data = df['listed_in'].dropna()

    # Split multiple genres and flatten
    genres = genre_data.str.split(', ').explode()

    # Get top 10 genres
    top_genres = genres.value_counts().head(10)

    # Plot
    plt.figure(figsize=(8,5))
    top_genres.plot(kind='bar')
    plt.title("Top 10 Genres on Netflix")
    plt.xlabel("Genre")
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.show()

else:
    print("❌ 'listed_in' column not found in dataset")
```

```python
print("INSIGHTS:")

print("1. Majority content type:", df['type'].value_counts().idxmax())
print("2. Peak release year:", df['release_year'].value_counts().idxmax())
print("3. Most content from:", df['country'].value_counts().idxmax())
print("4. Most common rating:", df['rating'].value_counts().idxmax())
```
